Opinion Classifier Based on the BERT Model

1. Environment setup
2. Definition of categories and establishment of their vector representations
3. Manual classification + Gradio interface

1. Configuracion del Entorno

```python
from sentence_transformers import SentenceTransformer

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity
import gradio as gr

model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
```

2. Definicion de categorias y establecimiento de sus vectores

```python
categories = {
    "positive":[
        "Fue un dia productivo.",
        "Estoy satisfecho.",
        "Muy buena calidad.",
        "El juego fue muy entretenido.",
        "Me diverti mucho jugando.",
        "Tuvo una nota perfecta.",
        "Mecanicas de juego muy satisfactorias.",
        "Juego perfecto.",
    ],
    "negative":[
        "No lo recomiendo, es muy de mala calidad.",
        "Fue una perdida de tiempo.",
        "El juego era aburrido y frustrante.",
        "Quede completamente decepcionado.",
        "Estoy insatisfecho.",
        "La jugabilidad era monotoma.",
        "Pesimo servicio.",
    ]
}

category_vectors = {}
for cat, examples in categories.items():
    embeddings = model.encode(examples)
    category_vectors[cat] = np.mean(embeddings, axis=0)
```

3. Manual de clasificacion + Interface de Gradio

```python
def classify_opinion(text):
    embedding = model.encode([text])[0]
    sims = {cat: cosine_similarity([embedding], [vec])[0][0] for cat, vec in category_vectors.items()}
    predicted = max(sims, key=sims.get)
    response = f" Opinion: {text}\n\n Classification: {predicted.upper()}\n\n Similarity:\n"
    for cat, score in sims.items():
        response += f"→ {cat}: {score:.4f}\n"
    return response
```

```python
gr.Interface(
    fn = classify_opinion,
    inputs = gr.Textbox(label = "Ingresar tu frase de opinion:"),
    outputs = gr.Textbox(label = "Resultado de la clasificación:"),
    title = "Clasificador de Opinión",
    description = "Escribe una opinion sobre el juego o producto, y el modelo evaluara si es positiva o negativa.",
    theme = "seafoam"
).launch()  #lanza la aplicacion
```
